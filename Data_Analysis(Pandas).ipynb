### Percentage of Developers who know Python per Country

To calculate the percentage of developers who know Python in each country, we first count the number of developers in each country who have indicated 'Python' in their 'LanguageHaveWorkedWith' response. Then, we divide this count by the total number of respondents for each country and multiply by 100.

```python
python_users_per_country = country_grp['LanguageHaveWorkedWith'].apply(lambda x: x.str.contains('Python').sum())
total_respondents_per_country = country_grp['Country'].count()

python_percentage_per_country = (python_users_per_country / total_respondents_per_country) * 100

# Sort the results in descending order
display(python_percentage_per_country.sort_values(ascending=False))
```
